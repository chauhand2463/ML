# Implementing Linear Regression From scratch

```python
import numpy as np;
import pandas as pd;
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('data.csv')
```

```python
def loss_function(m, b, points):
    x = points["Studytime"]
    y = points["score"]
    total_error = np.sum((y - (m * x + b)) ** 2)
    return total_error / float(len(points))


# this is loss function it tells us how far we are off from the fit line
```

```python
def gradient_descent(m_now, b_now, points, learning_rate):
    x = points["Studytime"]
    y = points["score"]
    n = len(points)

    predictions = m_now * x + b_now
    m_gradient = -(2 / n) * np.sum(x * (y - predictions))
    b_gradient = -(2 / n) * np.sum(y - predictions)

    m = m_now - ( m_gradient * learning_rate )
    b = b_now - ( b_gradient * learning_rate )
    return m, b
```

```python
m = 0
b = 0
L = 0.01
epochs = 1000

for i in range(epochs):
    if i % 50 == 0:
        print(f"epochs: {i}")
    m, b = gradient_descent(m, b, data, L)

print(m, b)
print("final loss:", loss_function(m, b, data))
```
