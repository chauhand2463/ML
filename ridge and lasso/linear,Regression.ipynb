# Linear regression example by califronia_housing dataset 

```python
import pandas as pd 
```

```python
from sklearn.datasets import fetch_california_housing
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = fetch_california_housing()
```

```python
df
```

```python
dataset = pd.DataFrame(df.data)
```

```python
dataset.columns = df.feature_names
```

```python
dataset.head()
```

```python
#  independent features and dependet fetures 
X = dataset
y = df.target
```

```python
# standardizing  the dataset cause output will better 

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train = scaler.fit_transform(X_train)  # type: ignore
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
## cross validation
from sklearn.model_selection import  cross_val_score
import numpy as np 
```

```python
regression = LinearRegression()
regression.fit(X_train, y_train)  
```

```python
mse = cross_val_score(regression,X_train,y_train,scoring = 'neg_mean_squared_error',cv = 5)
print("MSE for each fold:", mse)
print("Average MSE:", np.mean(mse))
```

```python
# preddiction on test data
reg_pred = regression.predict(X_test)
```

```python
reg_pred
```

```python
import seaborn as sns

sns.displot(reg_pred - y_test, kind="kde")
```

```python
from sklearn.metrics import r2_score 
score = r2_score(y_test, reg_pred)
score
```

# Ridge Regression with sklearn


```python
from sklearn.linear_model._ridge import Ridge
from sklearn.model_selection import GridSearchCV
```

```python
ridge_regressor = Ridge()
ridge_regressor
```

```python
parameters = {'alpha': [1,2,4,10,50,100,110]}
ridgecv = GridSearchCV(ridge_regressor, parameters, scoring='neg_mean_squared_error', cv=5)
ridgecv.fit(X_train, y_train)
```

```python
print(ridgecv.best_params_)
```

```python
print(ridgecv.best_score_)
```

```python
ridge_pred = ridgecv.predict(X_test)
```

```python
import seaborn as sns
sns.displot(ridge_pred-y_test , kind = 'kde')
```

```python
score = r2_score(ridge_pred,y_test)
score
```
